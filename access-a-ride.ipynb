```python
import pandas as pd
```

```python
df = pd.read_csv("MTA_Paratransit_Feb_24_May_26")
df
```

```python
# Find the number of unique providers
df["Provider_Type"].unique()
```

```python
# Extract month and date
## I asked Claude: "Using pandas, how can I find the total number of complaints per month from the attached dataset?"
df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.to_period("M")

# Total complaints per month
monthly_totals = df.groupby("Month")["Complaints"].sum().reset_index()

monthly_totals
```

```python
### To verify, I copied the MTA data for a random month (April 2024) into a spreadsheet and summed together the daily complaints to find that there were 3983. 
```

```python
# Define each provider 
primary = df[df["Provider_Type"] == "Primary"]
broker = df[df["Provider_Type"] == "Broker"]
supplemental = df[df["Provider_Type"] == "Supplemental"]
other = df[df["Provider_Type"] == "Other/Unidentified Provider"]
ehail = df[df["Provider_Type"] == "E-Hail"]
```

```python
# Filter monthly complaints by provider type
## I asked Claude: "How can I filter the monthly complaints by provider type (primary, broker, etc.)?

# Primary provider complaints per month
primary_monthly = (
    primary
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

primary_monthly
```

```python
### To verify, I copied the MTA data for a random month (January 2025) into a spreadsheet and used filters and a pivot table to find 735 primary provider complaints. 
```

```python
# Broker provider complaints per month
broker["Date"] = pd.to_datetime(broker["Date"])

broker_monthly = (
    broker
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

broker_monthly
```

```python
# Supplemental provider complaints per month
supplemental["Date"] = pd.to_datetime(supplemental["Date"])

supplemental_monthly = (
    supplemental
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

supplemental_monthly
```

```python
# Other provider complaints per month
other["Date"] = pd.to_datetime(other["Date"])

other_monthly = (
    other
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

other_monthly
```

```python
# Ehail provider complaints per month
ehail["Date"] = pd.to_datetime(ehail["Date"])

ehail["Date"] = pd.to_datetime(ehail["Date"])
ehail_monthly = (
    ehail
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

ehail_monthly
```

```python
# Find the number of unique complaint types
df["Category"].unique()
```

```python
# List the complaints by the number of each complaint type
df.groupby("Category")["Complaints"].sum().sort_values(ascending=False)
```

```python
# List the complaints by the number of each complaint type
df.groupby(["Provider_Type", "Category"])["Complaints"].sum()
```

```python
# Group and count complaints by month
## I asked Claude: "How can I sort the number of monthly complaints by both complaint type and provider type?"

# First, group complaints by month 
category_monthly = (
    df.groupby(["Month", "Category", "Provider_Type"])
    .size()
    .reset_index(name="Complaints")
)

# Filter the monthly complaints to obtain primary provider complaints
category_monthly[category_monthly["Provider_Type"] == "Primary"].sort_values("Complaints", ascending=False)
```

```python
# Broker provider complaints
category_monthly[category_monthly["Provider_Type"] == "Broker"].sort_values("Complaints", ascending=False)
```
