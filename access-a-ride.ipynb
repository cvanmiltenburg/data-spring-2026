```python
import pandas as pd
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/cvanmiltenburg/data-spring-2026/refs/heads/main/MTA_NYCT_Paratransit_Customer_Complaints__Beginning_February_2024_20260417.csv")
df
```

```python
# Find the number of unique providers
df["Provider_Type"].unique()
```

```python
primary = df[df["Provider_Type"] == "Primary"]
broker = df[df["Provider_Type"] == "Broker"]
supplemental = df[df["Provider_Type"] == "Supplemental"]
other = df[df["Provider_Type"] == "Other/Unidentified Provider"]
ehail = df[df["Provider_Type"] == "E-Hail"]
```

```python
# Change the date format and filter monthly complaints by provider type
primary["Date"] = pd.to_datetime(primary["Date"])

primary_monthly = (
    primary
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

primary_monthly
```

```python
broker["Date"] = pd.to_datetime(broker["Date"])

broker_monthly = (
    broker
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

broker_monthly
```

```python
supplemental["Date"] = pd.to_datetime(supplemental["Date"])

supplemental_monthly = (
    supplemental
    .groupby(pd.Grouper(key="Date", freq="ME"))["Complaints"]
    .sum()
    .reset_index()
)

supplemental_monthly
```
