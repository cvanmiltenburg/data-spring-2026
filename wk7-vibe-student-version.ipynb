````python
# Step 1: Load the CSV file into a DataFrame
nypd_summons = pd.read_csv('summons-15-25.csv')

# Step 2: Count the number of summons per borough, sorted descending by default
borough_counts = nypd_summons['BORO'].value_counts()

# Step 3: Convert the Series to a DataFrame for cleaner output
borough_counts = borough_counts.reset_index()

# Step 4: Rename columns for clarity (pandas 2.0+ uses 'count' as the default count column name)
borough_counts.columns = ['borough', 'summons_count']

# Step 5: (Optional) Drop rows where borough is missing/blank
borough_counts = borough_counts.dropna(subset=['borough'])

# Step 6: Display the result
print(borough_counts)
````
